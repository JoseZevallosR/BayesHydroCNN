# Datos

```python
# Definir rutas
depth_path = "D:/depths/TIF/DEPTHS"
friction_path = "D:/depths/TIF/FRICTION"

# Cargar imágenes y coordenadas
friction_images, x_coords, y_coords = load_tiff_images_with_coords(friction_path)
depth_images, _, _ = load_tiff_images_with_coords(depth_path)

# Normalización Min-Max de coordenadas
x_coords = (x_coords - np.min(x_coords)) / (np.max(x_coords) - np.min(x_coords))
y_coords = (y_coords - np.min(y_coords)) / (np.max(y_coords) - np.min(y_coords))

# Concatenar imágenes con coordenadas
X_data = np.concatenate([friction_images, x_coords[..., np.newaxis], y_coords[..., np.newaxis]], axis=-1)
y_data = depth_images

# División de datos en entrenamiento y prueba
X_train, X_val, y_train, y_val = train_test_split(X_data, y_data, test_size=0.2, random_state=42)
```

```python
correlation_map_train = correlation_weigths(X_train, y_train)
correlation_map_val = correlation_weigths(X_val, y_val)
correlation_map_train = jnp.nan_to_num(correlation_map_train, nan=0.0)
correlation_map_val = jnp.nan_to_num(correlation_map_val, nan=0.0)
```

# Cargando Modelo entrenado

```python
from flax.training import checkpoints
import os

# =======================
# Inicializar modelo base
# =======================
rng = jax.random.PRNGKey(0)
learning_rate = 1e-3
optimizer = optax.adam(learning_rate)

model = CNNModel(use_correlation=False)
params = model.init(rng, X_train[:1], get_correlation_map_batch(correlation_map_train, 1))
state = TrainState.create(apply_fn=model.apply, params=params, tx=optimizer)

# =======================
# Restaurar desde checkpoint
# =======================
ckpt_dir = os.path.abspath("D:/Proyectos_GitHub/adao_tests/notebooks/checkpoints_nn")
state = checkpoints.restore_checkpoint(ckpt_dir=ckpt_dir, target=state)

print("✅ Estado restaurado desde el checkpoint más reciente.")
```

```python
np.random.seed(42)
# Definir el tamaño del subset (50% de X_train)
subset_size = int(0.5 * X_train.shape[0])

# Seleccionar índices aleatorios
subset_indices = np.random.choice(X_train.shape[0], subset_size, replace=False)

# Crear el subset de X_train
X_train_subset = X_train[subset_indices]
y_train_subset = y_train[subset_indices]
correlation_map_train_subset = correlation_map_train[subset_indices]

# Obtener el batch size y la shape de salida esperada
batch_size = X_train.shape[0]
# Expandir y redimensionar el mapa de correlación para predicción
correlation_map_batch_train = get_correlation_map_batch(correlation_map_train_subset, batch_size)

# Aplicar modelo
y_pred_train_subset = state.apply_fn(state.params, X_train_subset, correlation_map_batch_train)
```

```python
# Obtener el batch size y la shape de salida esperada
batch_size = X_val.shape[0]

# Simula una pasada para saber la shape de salida esperada
sample_output = state.apply_fn(state.params, X_val[:1], jnp.zeros((1, 387, 450, 1)))
output_shape = sample_output.shape  # (1, 392, 456, 1) por ejemplo

# Expandir y redimensionar el mapa de correlación para predicción
correlation_map_batch_val = get_correlation_map_batch(correlation_map_val, batch_size)

# Aplicar modelo
y_pred_test = state.apply_fn(state.params, X_val, correlation_map_batch_val)
```

```python
# Usage
shapefile_path = "puntos_gp.shp"
data, _, selected_points = generate_samples_from_shp(X_train_subset, y_train_subset, y_pred_train_subset, shapefile_path)
plot_subplots(data, output_path='weigthed_subplots_depth_vs_friction_train.png')
```

```python
# Usage
shapefile_path = "puntos_gp.shp"
data, _, selected_points = generate_samples_from_shp(X_val, y_val, y_pred_test, shapefile_path)
plot_subplots(data, output_path='weigthed_subplots_depth_vs_friction_test.png')
```

```python
# Seleccionar 4 imágenes aleatorias del conjunto de validación
num_samples = 4
indices = np.random.choice(X_val.shape[0], num_samples, replace=False)

# Convertir de JAX tensor a NumPy para visualizar
y_pred_np = np.array(y_pred_test)  # Predicciones
y_test_np = np.array(y_val)  # Valores reales

# Calcular la diferencia entre la imagen real y la predicción
difference = y_test_np - y_pred_np

# Obtener los valores de rugosidad (n1, n2) para cada imagen seleccionada
roughness_values = get_n1_n2(X_val, indices)

# Crear la figura con 4 filas y 4 columnas (16 subplots en total)
fig, axes = plt.subplots(4, 4, figsize=(18, 16))

for i, idx in enumerate(indices):
    # Extraer imagen del conjunto de validación
    sample_image = X_val[idx]  # Forma esperada: (alto, ancho, 3)

    # Extraer el canal de fricción
    friction = sample_image[:, :, 0]  # Primer canal = fricción

    # Obtener los valores de rugosidad (n1, n2) para esta imagen
    n1, n2 = roughness_values[i]  # ✅ Cambiado de idx a i

    # Calcular la desviación estándar del error
    error_values = difference[idx, :, :, 0].flatten()  # Aplanar errores
    std_dev = np.std(error_values)  # Desviación estándar

    # Posiciones en los subplots
    row = i  # Fila de la imagen

    # 1️⃣ **Imagen Real**
    axes[row, 0].imshow(y_test_np[idx, :, :, 0], cmap="viridis")
    axes[row, 0].set_title(f"Simulated depth (n1={n1:.3f}, n2={n2:.3f})")
    axes[row, 0].axis("off")

    # 2️⃣ **Imagen Predicha**
    axes[row, 1].imshow(y_pred_np[idx, :, :, 0], cmap="viridis")
    axes[row, 1].set_title("Emulated depth")
    axes[row, 1].axis("off")

    # 3️⃣ **Diferencia entre Real y Predicho**
    im1 = axes[row, 2].imshow(difference[idx, :, :, 0], cmap="coolwarm", 
                              vmin=-np.max(abs(difference)), vmax=np.max(abs(difference)))
    axes[row, 2].set_title("Error spatial map")
    fig.colorbar(im1, ax=axes[row, 2], fraction=0.046, pad=0.04)
    axes[row, 2].axis("off")

    # 4️⃣ **Histograma de Errores**
    axes[row, 3].hist(error_values, bins=50, color='blue', edgecolor='black', alpha=0.7)
    axes[row, 3].set_xlabel("Error")
    axes[row, 3].set_ylabel("Frequency")
    axes[row, 3].set_title("Error histogram")
    axes[row, 3].grid(True)

# Ajustar diseño
plt.tight_layout()

# Guardar la figura
plt.savefig("subplots_results.png", dpi=300)

# Mostrar la figura
plt.show()
```
