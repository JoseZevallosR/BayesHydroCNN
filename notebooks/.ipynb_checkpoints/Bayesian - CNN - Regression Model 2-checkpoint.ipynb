```python
# Definir rutas
depth_path = "D:/depths/TIF/DEPTHS"
friction_path = "D:/depths/TIF/FRICTION"

# Cargar imágenes y coordenadas
friction_images, x_coords, y_coords = load_tiff_images_with_coords(friction_path)
depth_images, _, _ = load_tiff_images_with_coords(depth_path)

# Normalización Min-Max de coordenadas
x_coords = (x_coords - np.min(x_coords)) / (np.max(x_coords) - np.min(x_coords))
y_coords = (y_coords - np.min(y_coords)) / (np.max(y_coords) - np.min(y_coords))

# Concatenar imágenes con coordenadas
X_data = np.concatenate([friction_images, x_coords[..., np.newaxis], y_coords[..., np.newaxis]], axis=-1)
y_data = depth_images

# División de datos en entrenamiento y prueba
X_train, X_val, y_train, y_val = train_test_split(X_data, y_data, test_size=0.2, random_state=42)
correlation_map_train = correlation_weigths(X_train, y_train)
correlation_map_val = correlation_weigths(X_val, y_val)
correlation_map_train = jnp.nan_to_num(correlation_map_train, nan=0.0)
correlation_map_val = jnp.nan_to_num(correlation_map_val, nan=0.0)
```

```python
from flax.training import checkpoints
import os

# =======================
# Inicializar modelo base
# =======================
rng = jax.random.PRNGKey(0)
learning_rate = 1e-3
optimizer = optax.adam(learning_rate)

model = CNNModel(use_correlation=False)
params = model.init(rng, X_train[:1], get_correlation_map_batch(correlation_map_train, 1))
#params = model.init(rng, dummy_input, get_correlation_map_batch(correlation_map_train, 1))
state = TrainState.create(apply_fn=model.apply, params=params, tx=optimizer)

# =======================
# Restaurar desde checkpoint
# =======================
ckpt_dir = os.path.abspath("D:/Proyectos_GitHub/adao_tests/notebooks/checkpoints_nn")
state = checkpoints.restore_checkpoint(ckpt_dir=ckpt_dir, target=state)

print("✅ Estado restaurado desde el checkpoint más reciente.")
```

```python
sigma = 0.1 
ruido = np.random.normal(loc=0, scale=sigma, size=(387, 450, 1))
# 🔹 **Cargar datos asegurando `float64`**
x_cov = X_train[0,:,:,:]
true_image = y_val[2,:,:,:] + ruido
correlation_map = jnp.array(np.load("correlation_map.npy"))
# ✅ **Asegurar que `true_image` tiene la forma correcta**
true_image = np.squeeze(true_image)  # Elimina dimensiones de tamaño 1 si exist
```

```python
import numpy as np

# Umbral de correlación
threshold = 0.6

# Asegurarse de que correlation_map sea una matriz numpy si es jax
correlation_map_np = np.array(correlation_map)

# Obtener la máscara booleana de los píxeles que superan el umbral
mask = correlation_map_np > threshold

# Extraer las coordenadas (y, x) donde se cumple la condición
coords_y, coords_x = np.where(mask)

# Extraer los valores de profundidad observados en esas coordenadas
observed_values = true_image[coords_y, coords_x]
```

```python
with pm.Model() as model:
    # Priors
    n1_prior = pm.Uniform("n1", lower=0.01, upper=0.085)
    n2_prior = pm.Uniform("n2", lower=0.01, upper=0.085)
    prec = pm.Gamma("prec", alpha=0.5, beta=750.0)
    sigma = pm.Deterministic("sigma", 1 / pt.sqrt(prec))

    # Tensor de entrada
    x_cov_t = pt.as_tensor_variable(x_cov)

    # Inicializar la operación personalizada con el estado del modelo
    cnn_op = CNNPredictOp(model_state=state, correlation_map=correlation_map)

    # Predicción
    depth_mean = cnn_op(x_cov_t, n1_prior, n2_prior)
    depth_full = pt.reshape(depth_mean, true_image.shape)

    # Extraer solo los puntos muestreados
    depth_subset = depth_full[coords_y, coords_x]

    # Observación parcial
    pm.Normal("observed", mu=depth_subset, sigma=sigma, observed=observed_values)

    # Configurar el muestreador Slice
    step = pm.Slice()
    # Muestreo
    trace = pm.sample(
        draws=100,
        tune=20,
        chains=4,
        cores=4,
        step=step,
        idata_kwargs={"log_likelihood": False}  # Evita guardar log_likelihood
    )
```

```python
# Obtener resumen estadístico del trazado
summary = az.summary(trace, var_names=["n1", "n2","sigma"])

# Mostrar el resumen
summary
```

# MvNormal

```python
def sample_observed_values(correlation_map, true_image, threshold_1=0.5, threshold_2=0.75, n_samples=100):
    """
    Devuelve un array de valores observados (true_image) en coordenadas seleccionadas
    aleatoriamente según dos umbrales de correlación, junto con las coordenadas (y, x) correspondientes.

    Parámetros:
    - correlation_map: matriz 2D con valores de correlación.
    - true_image: matriz 2D con los valores reales de profundidad.
    - threshold_1: primer umbral de correlación (por defecto 0.5).
    - threshold_2: segundo umbral de correlación (por defecto 0.75).
    - n_samples: número de muestras aleatorias por umbral (por defecto 100).

    Retorna:
    - observed_values: array de tamaño (2 * n_samples,) con los valores combinados.
    - coords_y: array con coordenadas y (filas) combinadas.
    - coords_x: array con coordenadas x (columnas) combinadas.
    """
    correlation_map_np = np.array(correlation_map)

    # Máscaras para cada umbral
    mask_1 = correlation_map_np >= threshold_1
    mask_2 = correlation_map_np >= threshold_2

    # Coordenadas que cumplen cada condición
    coords_y_1, coords_x_1 = np.where(mask_1)
    coords_y_2, coords_x_2 = np.where(mask_2)

    # Validación
    if len(coords_y_1) < n_samples or len(coords_y_2) < n_samples:
        raise ValueError("No hay suficientes píxeles que cumplen los criterios de correlación.")

    # Muestreo aleatorio sin reemplazo
    idx_1 = np.random.choice(len(coords_y_1), n_samples, replace=False)
    idx_2 = np.random.choice(len(coords_y_2), n_samples, replace=False)

    # Coordenadas seleccionadas
    selected_y_1, selected_x_1 = coords_y_1[idx_1], coords_x_1[idx_1]
    selected_y_2, selected_x_2 = coords_y_2[idx_2], coords_x_2[idx_2]

    # Valores observados combinados
    observed_values_1 = true_image[selected_y_1, selected_x_1]
    observed_values_2 = true_image[selected_y_2, selected_x_2]

    observed_values = np.concatenate([observed_values_1, observed_values_2])
    coords_y = np.concatenate([selected_y_1, selected_y_2])
    coords_x = np.concatenate([selected_x_1, selected_x_2])

    return observed_values, coords_y, coords_x


observed_values,coords_y, coords_x = sample_observed_values(correlation_map, true_image, n_samples=20)
```

```python
import pymc as pm
import pytensor.tensor as pt
import numpy as np

with pm.Model() as model:
    # Priors
    n1 = pm.Uniform("n1", lower=0.01, upper=0.085)
    n2 = pm.Uniform("n2", lower=0.01, upper=0.085)
    prec = pm.Gamma("prec", alpha=0.5, beta=750.0)
    sigma = pm.Deterministic("sigma", 1 / pt.sqrt(prec))

    # Tensor de entrada
    x_cov_t = pt.as_tensor_variable(x_cov)

    # Emulador CNN
    cnn_op = CNNPredictOp(model_state=state, correlation_map=correlation_map)
    depth_mean = cnn_op(x_cov_t, n1, n2)
    depth_full = pt.reshape(depth_mean, true_image.shape)
    depth_subset = depth_full[coords_y, coords_x]  # μ para MvNormal

    # Coordenadas (para el kernel)
    X_coords = np.stack([coords_y, coords_x], axis=1)

    # Kernel GP (Ej: ExpQuad)
    ℓ = pm.Gamma("ℓ", alpha=2, beta=1)
    η = pm.HalfNormal("η", sigma=1)

    from pymc.gp.cov import ExpQuad
    cov_func = η**2 * ExpQuad(input_dim=2, ls=ℓ)

    # Dimensión
    n_obs = observed_values.shape[0]

    # Matriz de covarianza + jitter
    Sigma = cov_func(X_coords) + 1e-6 * pt.eye(n_obs)

    # Cov total (ruido adicional)
    cov_total = Sigma + sigma**2 * pt.eye(n_obs)

    # Observación multivariada
    pm.MvNormal("observed", mu=depth_subset, cov=cov_total, observed=observed_values)

    # Muestreo
    step = pm.Slice()
    trace = pm.sample(
        draws=1000,
        tune=500,
        chains=4,
        cores=4,
        step=step,
        idata_kwargs={"log_likelihood": False}
    )

```

```python
# Obtener resumen estadístico del trazado
summary = az.summary(trace, var_names=["n1", "n2","sigma"])

# Mostrar el resumen
summary
```

# Inadecuación

```python
import pymc as pm
import pytensor.tensor as pt
import numpy as np
from pymc.gp.cov import ExpQuad

with pm.Model() as model:
    # Priors for parameters
    n1 = pm.Uniform("n1", lower=0.01, upper=0.085)
    n2 = pm.Uniform("n2", lower=0.01, upper=0.085)
    prec = pm.Gamma("prec", alpha=0.5, beta=750.0)
    sigma = pm.Deterministic("sigma", 1 / pt.sqrt(prec))  # observational noise

    # Input tensor
    x_cov_t = pt.as_tensor_variable(x_cov)

    # CNN emulator (η)
    cnn_op = CNNPredictOp(model_state=state, correlation_map=correlation_map)
    depth_mean = cnn_op(x_cov_t, n1, n2)
    depth_full = pt.reshape(depth_mean, true_image.shape)
    depth_subset = depth_full[coords_y, coords_x]  # η(x, θ)

    # Observation coordinates
    X_coords = np.stack([coords_y, coords_x], axis=1)

    # GP for model inadequacy δ(x) ~ GP(0, Σ)
    ℓ_delta = pm.Gamma("ℓ_delta", alpha=2, beta=1)
    η_delta = pm.HalfNormal("η_delta", sigma=1)
    cov_delta = η_delta**2 * ExpQuad(input_dim=2, ls=ℓ_delta)

    gp_inadequacy = pm.gp.Latent(cov_func=cov_delta)
    delta = gp_inadequacy.prior("delta", X=X_coords)

    # Combined mean: emulator output + model discrepancy
    mu_total = depth_subset + delta

    # Final likelihood with noise
    n_obs = observed_values.shape[0]
    pm.MvNormal(
        "observed",
        mu=mu_total,
        cov=sigma**2 * pt.eye(n_obs),
        observed=observed_values
    )

    # Inference
    step = pm.Slice()
    trace = pm.sample(
        draws=500,
        tune=100,
        chains=4,
        cores=4,
        step=step,
        idata_kwargs={"log_likelihood": False}
    )
```

```python
import arviz as az
import matplotlib.pyplot as plt
import numpy as np

# Valores reales
true_n1 = 0.027
true_n2 = 0.057

# Rango del prior uniforme
prior_lower = 0.01
prior_upper = 0.085
prior_height = 1 / (prior_upper - prior_lower)

# Trazas
fig = az.plot_trace(trace, var_names=["n1", "n2", "sigma"], figsize=(12, 8))

# Agregar densidad del prior uniforme en los histogramas (columna 0)
# n1
ax_n1 = fig[0, 0]
x_vals = np.linspace(prior_lower, prior_upper, 100)
y_vals = np.ones_like(x_vals) * prior_height
ax_n1.plot(x_vals, y_vals, color='green', linestyle='-', linewidth=1.5, label="Prior density")
ax_n1.axvline(true_n1, color="red", linestyle="--", label="True n1")
ax_n1.axvline(prior_lower, color="gray", linestyle=":", label="Prior lower")
ax_n1.axvline(prior_upper, color="gray", linestyle=":", label="Prior upper")
ax_n1.set_ylabel("Density")
# n2
ax_n2 = fig[1, 0]
ax_n2.plot(x_vals, y_vals, color='green', linestyle='-', linewidth=1.5, label="Prior density")
ax_n2.axvline(true_n2, color="red", linestyle="--", label="True n2")
ax_n2.axvline(prior_lower, color="gray", linestyle=":", label="Prior lower")
ax_n2.axvline(prior_upper, color="gray", linestyle=":", label="Prior upper")
ax_n2.set_ylabel("Density")
# Solo una leyenda (en el primero)
ax_n1.legend()

# n2
ax_sigma2 = fig[2, 0]
ax_sigma2.set_ylabel("Density")

plt.tight_layout()
# ----- Guardar la figura -----
output_path = "trace_plot_with_prior_posterior_true.png"
plt.tight_layout()
plt.savefig(output_path, dpi=300, bbox_inches='tight')
plt.show()

print(f"✅ Imagen guardada en: {output_path}")
```
