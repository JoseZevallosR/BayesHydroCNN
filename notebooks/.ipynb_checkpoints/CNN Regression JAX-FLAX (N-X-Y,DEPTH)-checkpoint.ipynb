```python
import jax.numpy as jnp
import flax.linen as nn

import os
import jax
import jax.numpy as jnp
import flax.linen as nn
import optax
import rasterio
import numpy as np
import pickle
from sklearn.model_selection import train_test_split
from flax.training import train_state
import matplotlib.pyplot as plt

# ============================
# Cargar imágenes y coordenadas
# ============================
def load_tiff_images_with_coords(folder):
    images, x_coords_list, y_coords_list = [], [], []
    files = sorted(os.listdir(folder))

    for file in files:
        if file.endswith(".tiff") or file.endswith(".tif"):
            file_path = os.path.join(folder, file)
            with rasterio.open(file_path) as src:
                image = src.read(1).astype(np.float64)
                image = np.round(image, 4)
                image = np.expand_dims(image, axis=-1)
                images.append(image)

                # Obtener coordenadas X e Y
                height, width = src.height, src.width
                transform = src.transform
                x_coords = np.array([transform * (x, 0) for x in range(width)])[:, 0]
                x_coords = np.tile(x_coords[np.newaxis, :], (height, 1))
                y_coords = np.array([[transform * (0, y)][0][1] for y in range(height)])
                y_coords = np.tile(y_coords[:, np.newaxis], (1, width))

                x_coords_list.append(x_coords)
                y_coords_list.append(y_coords)

    return (
        np.array(images, dtype=np.float64),
        np.array(x_coords_list, dtype=np.float64),
        np.array(y_coords_list, dtype=np.float64)
    )

```

```python
# Definir rutas
depth_path = "D:/depths/TIF/DEPTHS"
friction_path = "D:/depths/TIF/FRICTION"

# Cargar imágenes y coordenadas
friction_images, x_coords, y_coords = load_tiff_images_with_coords(friction_path)
depth_images, _, _ = load_tiff_images_with_coords(depth_path)

# Normalización Min-Max de coordenadas
x_coords = (x_coords - np.min(x_coords)) / (np.max(x_coords) - np.min(x_coords))
y_coords = (y_coords - np.min(y_coords)) / (np.max(y_coords) - np.min(y_coords))

# Concatenar imágenes con coordenadas
X_data = np.concatenate([friction_images, x_coords[..., np.newaxis], y_coords[..., np.newaxis]], axis=-1)
y_data = depth_images

# División de datos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X_data, y_data, test_size=0.2, random_state=42)

# Convertir a formato JAX (float32 para compatibilidad)
X_train = jnp.array(X_train, dtype=jnp.float32)
X_test = jnp.array(X_test, dtype=jnp.float32)
y_train = jnp.array(y_train, dtype=jnp.float32)
y_test = jnp.array(y_test, dtype=jnp.float32)
```

```python
def weighted_mse_loss(y_true, y_pred, weight_map):
    """
    Función de pérdida MSE ponderada que da más peso a las regiones con mayores diferencias.
    
    Parámetros:
        y_true (jnp.array): Valores reales de la profundidad.
        y_pred (jnp.array): Valores predichos por la red.
        weight_map (jnp.array): Mapa de pesos que enfatiza zonas importantes.
    
    Retorna:
        Pérdida MSE ponderada.
    """
    error = jnp.square(y_true - y_pred)  # MSE normal
    weight_map = jnp.expand_dims(weight_map, axis=(0, -1))  # Expande a (1, 387, 450, 1)
    weighted_error = error * weight_map  # Aplica el peso a cada pixel
    return jnp.mean(weighted_error)

# def create_weight_map(shape, center_strength=3.0):
#     """
#     Crea un mapa de pesos donde el centro tiene valores más altos.
    
#     Parámetros:
#         shape (tuple): Dimensiones de la imagen (height, width).
#         center_strength (float): Factor que amplifica el peso en el centro.
    
#     Retorna:
#         Mapa de pesos (jnp.array) con mayor énfasis en la región central.
#     """
#     h, w = shape
#     y, x = jnp.ogrid[:h, :w]
    
#     center_y, center_x = h // 2, w // 2  # Centro de la imagen
#     weight_map = jnp.exp(-((y - center_y)**2 / (h/4)**2 + (x - center_x)**2 / (w/4)**2))  # Gaussiana
#     weight_map = 1 + center_strength * (1 - weight_map)  # Ajusta la escala

#     return weight_map

def create_adaptive_weight_map(error_map, base_weight=1.0, max_weight=5.0):
    """
    Crea un mapa de pesos basado en el error en cada píxel.

    Parámetros:
        error_map (jnp.array): Mapa de errores entre la predicción y la verdad.
        base_weight (float): Peso mínimo en las regiones con menor error.
        max_weight (float): Peso máximo en las regiones con mayor error.

    Retorna:
        Mapa de pesos (jnp.array) con mayor énfasis donde el error es alto.
    """
    norm_error = (error_map - error_map.min()) / (error_map.max() - error_map.min() + 1e-8)  # Normaliza el error [0,1]
    weight_map = base_weight + norm_error * (max_weight - base_weight)  # Ajusta el rango de pesos

    return weight_map
```

```python
# ============================
# Definir el modelo en Flax
# ============================
class CNNModel(nn.Module):
    @nn.compact
    def __call__(self, x):
        x = nn.Conv(features=32, kernel_size=(5, 5), padding="SAME")(x)  # ⬅️ Cambiado a (5,5)
        x = nn.relu(x)
        x = nn.max_pool(x, window_shape=(2, 2), strides=(2, 2), padding="SAME")

        x = nn.Conv(features=64, kernel_size=(5, 5), padding="SAME")(x)  # ⬅️ Cambiado a (5,5)
        x = nn.relu(x)
        x = nn.max_pool(x, window_shape=(2, 2), strides=(2, 2), padding="SAME")

        x = nn.Conv(features=128, kernel_size=(5, 5), padding="SAME", kernel_dilation=(2, 2))(x)  # ⬅️ Cambiado
        x = nn.relu(x)
        x = nn.max_pool(x, window_shape=(2, 2), strides=(2, 2), padding="SAME")

        x = nn.Conv(features=128, kernel_size=(5, 5), padding="SAME", kernel_dilation=(2, 2))(x)  # ⬅️ Cambiado
        x = nn.relu(x)
        x = jnp.repeat(jnp.repeat(x, 2, axis=1), 2, axis=2)  # Simula UpSampling2D

        x = nn.Conv(features=64, kernel_size=(5, 5), padding="SAME")(x)  # ⬅️ Cambiado
        x = nn.relu(x)
        x = jnp.repeat(jnp.repeat(x, 2, axis=1), 2, axis=2)

        x = nn.Conv(features=32, kernel_size=(5, 5), padding="SAME")(x)  # ⬅️ Cambiado
        x = nn.relu(x)
        x = jnp.repeat(jnp.repeat(x, 2, axis=1), 2, axis=2)

        x = nn.Conv(features=1, kernel_size=(1, 1), padding="SAME")(x)

        # Simulación de Cropping2D
        x = x[:, 2:-3, 3:-3, :]
        return x
        
def create_adaptive_weight_map(y_true, y_pred, base_weight=1.0, max_weight=5.0):
    """
    Crea un mapa de pesos basado en el error en cada píxel.
    """
    error_map = jnp.abs(y_true - y_pred)
    norm_error = jnp.abs(error_map - jnp.mean(error_map)) / (jnp.std(error_map) + 1e-8)
    weight_map = base_weight + norm_error * (max_weight - base_weight)
    return jnp.broadcast_to(weight_map, y_true.shape)  # Asegura que tenga la misma forma

def weighted_mse_loss(y_true, y_pred, weight_map):
    """
    Función de pérdida MSE ponderada.
    """
    error = jnp.square(y_true - y_pred)
    weighted_error = error * weight_map  # Ahora weight_map tiene las dimensiones correctas
    return jnp.mean(weighted_error)

def loss_fn(params, apply_fn, x, y):
    predictions = apply_fn(params, x)
    weight_map = create_adaptive_weight_map(y, predictions)
    return weighted_mse_loss(y, predictions, weight_map)
    

        
import jax.numpy as jnp
import flax.linen as nn

# ============================
# Definir el estado de entrenamiento
# ============================
class TrainState(train_state.TrainState):
    pass

# ============================
# Crear el mapa de pesos antes del entrenamiento
# ============================
#image_shape = (387, 450)  # Tamaño de las imágenes
#weight_map = create_weight_map(image_shape)  # Llamamos a la función para generar pesos
# ============================
# Definir la función de pérdida
# ============================
# def loss_fn(params, apply_fn, x, y):
#     predictions = apply_fn(params, x)  # Se pasa directamente `params`
#     return jnp.mean((predictions - y) ** 2)
# def loss_fn(params, apply_fn, x, y):
#     predictions = apply_fn(params, x)  # Se pasa directamente `params`
#     return weighted_mse_loss(y, predictions, weight_map)  # Ahora usa la versión ponderada
# ============================
# Función de entrenamiento
# ============================
@jax.jit
def train_step(state, batch_x, batch_y):
    loss, grads = jax.value_and_grad(loss_fn)(state.params, state.apply_fn, batch_x, batch_y)
    state = state.apply_gradients(grads=grads)
    return state, loss
# ============================
# Inicializar modelo y optimizador
# ============================
rng = jax.random.PRNGKey(0)
model = CNNModel()
#model = UNetRegression()
learning_rate = 1e-3
optimizer = optax.adam(learning_rate)
```

```python
params = model.init(rng, X_train[:1])  # Inicializar modelo con una muestra

state = TrainState.create(apply_fn=model.apply, params=params, tx=optimizer)

# ============================
# Entrenar el modelo
# ============================
epochs = 100
batch_size = 16
num_batches = X_train.shape[0] // batch_size

loss_history = []

for epoch in range(epochs):
    total_loss = 0.0

    for i in range(num_batches):
        batch_x = X_train[i * batch_size:(i + 1) * batch_size]
        batch_y = y_train[i * batch_size:(i + 1) * batch_size]

        state, loss = train_step(state, batch_x, batch_y)
        total_loss += loss

    avg_loss = total_loss / num_batches
    loss_history.append(avg_loss)

    print(f"Epoch {epoch+1}, Loss: {avg_loss:.5f}")

    # Early stopping (si la pérdida no mejora en 3 épocas consecutivas)
    if len(loss_history) > 3 and loss_history[-1] >= loss_history[-3]:
        print("Stopping early due to no improvement.")
        break

# ============================
# Visualización del entrenamiento
# ============================
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Seleccionar una imagen aleatoria del conjunto de prueba
idx = np.random.randint(0, X_test.shape[0])

# Convertir de JAX tensor a NumPy para visualizar
y_pred_np = np.array(y_pred)  # Convertir la predicción a NumPy
y_test_np = np.array(y_test)  # Convertir el valor real a NumPy

# Calcular la diferencia entre la imagen real y la predicción
difference = y_test_np - y_pred_np

# Seleccionar la imagen del conjunto de prueba
sample_image = X_test[idx]  # Forma esperada: (alto, ancho, 3)

# Extraer el canal de fricción
friction = sample_image[:, :, 0]  # Primer canal = fricción

# Calcular la desviación estándar del error
error_values = difference[idx, :, :, 0].flatten()  # Aplanar los valores
std_dev = np.std(error_values)  # Calcular desviación estándar

# Crear la figura con 6 subgráficos (2 filas, 3 columnas)
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# Imagen de fricción con colorbar
im_friction = axes[0, 0].imshow(friction, cmap="gray")
axes[0, 0].set_title("Friction Map")
axes[0, 0].axis("off")
fig.colorbar(im_friction, ax=axes[0, 0], fraction=0.046, pad=0.04)

# Imagen real
axes[0, 1].imshow(y_test_np[idx, :, :, 0], cmap="viridis")
axes[0, 1].set_title("Real Depth")
axes[0, 1].axis("off")

# Imagen predicha
axes[0, 2].imshow(y_pred_np[idx, :, :, 0], cmap="viridis")
axes[0, 2].set_title("Predicted Depth")
axes[0, 2].axis("off")

# Diferencia con rango máximo absoluto
im1 = axes[1, 0].imshow(difference[idx, :, :, 0], cmap="coolwarm", 
                        vmin=-np.max(abs(difference)), vmax=np.max(abs(difference)))
axes[1, 0].set_title("Diferencia (True - Predicho)")
fig.colorbar(im1, ax=axes[1, 0], fraction=0.046, pad=0.04)
axes[1, 0].axis("off")

# Histograma de errores
axes[1, 1].hist(error_values, bins=50, color='blue', edgecolor='black', alpha=0.7)
axes[1, 1].set_xlabel("Error")
axes[1, 1].set_ylabel("Frecuencia")
axes[1, 1].set_title("Histograma de Errores")
axes[1, 1].grid(True)

# Diferencia con escala de desviación estándar
im2 = axes[1, 2].imshow(difference[idx, :, :, 0], cmap="coolwarm", vmin=-2*std_dev, vmax=2*std_dev)
axes[1, 2].set_title("Diferencias con Escala ± 2 * Desviación Estándar")
fig.colorbar(im2, ax=axes[1, 2], fraction=0.046, pad=0.04)
axes[1, 2].axis("off")

# Ajustar diseño
plt.tight_layout()
plt.show()
```

```python
# Calcular el error total en todo el conjunto
total_difference = y_test_np - y_pred_np
all_errors = total_difference[:, :, :, 0].flatten()

# Desviación estándar global del error
std_global = np.std(all_errors)

# Otras métricas útiles:
mae = np.mean(np.abs(all_errors))  # Mean Absolute Error
rmse = np.sqrt(np.mean(all_errors**2))  # Root Mean Squared Error
medae = np.median(np.abs(all_errors))  # Median Absolute Error

print(f"Desviación estándar global del error: {std_global:.4f}")
print(f"MAE (Error absoluto medio): {mae:.4f}")
print(f"RMSE (Raíz del error cuadrático medio): {rmse:.4f}")
print(f"MedAE (Error absoluto mediano): {medae:.4f}")
```

```python
# Calcular métricas por píxel
mae_pixel = np.mean(np.abs(y_test_np - y_pred_np), axis=0)[:, :, 0]  # MAE por píxel
rmse_pixel = np.sqrt(np.mean((y_test_np - y_pred_np) ** 2, axis=0))[:, :, 0]  # RMSE por píxel
medae_pixel = np.median(np.abs(y_test_np - y_pred_np), axis=0)[:, :, 0]  # MedAE por píxel
std_pixel = np.std(y_test_np - y_pred_np, axis=0)[:, :, 0]  # STD del error por píxel

# Crear figura y subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle("Distribución de Errores por Píxel", fontsize=16, fontweight="bold")

# Definir métricas y configuraciones
metrics = [
    ("MAE (Error Absoluto Medio)", mae_pixel, "inferno"),
    ("RMSE (Raíz del Error Cuadrático Medio)", rmse_pixel, "magma"),
    ("MedAE (Error Absoluto Mediano)", medae_pixel, "plasma"),
    ("STD del Error (Variabilidad del Error)", std_pixel, "viridis")
]

# Rango máximo común para normalizar colores y comparación
vmax_common = max(np.max(mae_pixel), np.max(rmse_pixel), np.max(medae_pixel))

for ax, (title, data, cmap) in zip(axes.flat, metrics):
    im = ax.imshow(data, cmap=cmap, vmin=0, vmax=vmax_common if "STD" not in title else None)
    ax.set_title(title, fontsize=12)
    ax.axis("off")

    # Barra de color con etiqueta
    cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    cbar.set_label("Valor del Error", fontsize=10)

# Ajustar diseño
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
